# 1. Data Understanding

```python
import pandas as pd
from pathlib import Path
```

```python
# Define the source data directory
data_path = Path("../data/raw")
```

## orders

> **-** 8 orders marked as delivered with no delivery date <br>
> **-** 6 orders marked as canceled that include a delivery date <br>
> **-** 166 instances where a shipment was handed over to the carrier before the purchase date <br>
> **-** 23 instances where a shipment was delivered before being handed over to the carrier <br>

```python
# Load the orders dataset
orders = pd.read_csv(data_path / "olist_orders_dataset.csv")
```

```python
print(orders.shape)
display(orders.head())
orders.info()
```

```python
print("Unique orders:", orders["order_id"].nunique())
```

```python
display(orders["order_status"].value_counts())
```

```python
# Count missing delivery dates by order status
display(pd.crosstab(orders["order_status"], orders["order_delivered_customer_date"].isna()))
```

```python
# Check delivered orders with missing delivery dates
display(orders.loc[(orders["order_status"] == "delivered") & (orders["order_delivered_customer_date"].isna())])
```

```python
# Check canceled orders with recorded delivery dates
display(orders.loc[(orders["order_status"] == "canceled") & (orders["order_delivered_customer_date"].notna())])
```

```python
# Define columns containing dates
date_cols = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

# Convert date columns to datetime
orders[date_cols] = orders[date_cols].apply(pd.to_datetime)

# Verify data types
orders[date_cols].dtypes
```

> Purchase → Approval → Carrier → Customer Delivery

```python
# Check orders approved before purchase
print("Approval before purchase:", (orders["order_approved_at"] < orders["order_purchase_timestamp"]).sum())
```

```python
# Check orders shipped before purchase
print("Shipping before purchase:",(orders["order_delivered_carrier_date"] < orders["order_purchase_timestamp"]).sum())
```

```python
# Select orders shipped before purchase
shipping_errors = orders[orders["order_delivered_carrier_date"] < orders["order_purchase_timestamp"]]

# Display sample records
display(shipping_errors[[
    "order_status",
    "order_purchase_timestamp",
    "order_delivered_carrier_date"
]].head(10))
```

```python
# Check orders delivered before shipping
print("Delivery before shipping:", (orders["order_delivered_customer_date"] < orders["order_delivered_carrier_date"]).sum())
```

```python
# Select orders delivered before shipping
delivery_errors = orders[orders["order_delivered_customer_date"] < orders["order_delivered_carrier_date"]]

# Display sample records
display(delivery_errors[[
    "order_status",
    "order_delivered_carrier_date",
    "order_delivered_customer_date"
]].head(10))
```

```python
# Check orders delivered before purchase
print("Delivery before purchase:", (orders["order_delivered_customer_date"] < orders["order_purchase_timestamp"]).sum())
```

```python
# Calculate time differences in hours
shipping_gap = (
    shipping_errors["order_purchase_timestamp"] -
    shipping_errors["order_delivered_carrier_date"]
).dt.total_seconds() / 3600

delivery_gap = (
    delivery_errors["order_delivered_carrier_date"] -
    delivery_errors["order_delivered_customer_date"]
).dt.total_seconds() / 3600

# Summarize shipping date inconsistencies
print("Shipping before purchase (hours):")
print(shipping_gap.describe().round(2))
print("Cases exceeding 24 hours:", (shipping_gap > 24).sum())

# Summarize delivery date inconsistencies
print("\nDelivery before shipping (hours):")
print(delivery_gap.describe().round(2))
print("Cases exceeding 24 hours:", (delivery_gap > 24).sum())
```

## other tables

```python
# Inspect the structure of the remaining datasets
for file in sorted(data_path.glob("*.csv")):

    if file.name == "olist_orders_dataset.csv":
        continue

    df = pd.read_csv(file)

    print(f"\nDataset: {file.name}")
    print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
    print("Column names:", df.columns.tolist())
```

```python
# Define candidate primary keys for each dataset
keys = {
    "olist_customers_dataset.csv": ["customer_id"],
    "olist_order_items_dataset.csv": ["order_id", "order_item_id"],
    "olist_order_payments_dataset.csv": ["order_id", "payment_sequential"],
    "olist_order_reviews_dataset.csv": ["review_id"],
    "olist_products_dataset.csv": ["product_id"],
    "olist_sellers_dataset.csv": ["seller_id"],
    "product_category_name_translation.csv": ["product_category_name"]
}

# Check duplicate and missing key values
for file, columns in keys.items():
    df = pd.read_csv(data_path / file)

    duplicates = df.duplicated(subset=columns).sum()
    missing = df[columns].isna().any(axis=1).sum()

    print(f"{file}")
    print(f"Duplicate keys: {duplicates}, Missing keys: {missing}\n")
```

> Where do the duplicate `review_id` keys come from?

```python
# Load the reviews dataset
reviews = pd.read_csv(data_path / "olist_order_reviews_dataset.csv")
```

```python
# Check duplicates of the combined review and order IDs
print("Duplicate review_id + order_id:", reviews.duplicated(["review_id", "order_id"]).sum())

# Select all records with repeated review IDs
duplicate_reviews = reviews[reviews.duplicated("review_id", keep=False)].sort_values("review_id")

# Check how many repeated review IDs refer to different orders
print("Review IDs linked to multiple orders:", (duplicate_reviews.groupby("review_id")["order_id"].nunique() > 1).sum())

# Display examples of repeated review IDs
display(duplicate_reviews[[
    "review_id",
    "order_id",
    "review_score",
    "review_creation_date"
]].head(12))
```

> Can a single `order_id` have multiple reviews?

```python
# Count reviews per order
review_counts = reviews["order_id"].value_counts()

# Check orders with multiple reviews
print("Orders with multiple reviews:", (review_counts > 1).sum())
print("Maximum reviews per order:", review_counts.max())

# Check missing order identifiers
print("Missing order IDs:", reviews["order_id"].isna().sum())

# Display examples of orders with multiple reviews
multiple_reviews = reviews[reviews["order_id"].isin(review_counts[review_counts > 1].index)]

display(multiple_reviews[["order_id", "review_id", "review_score"]].sort_values("order_id").head(10))
```
